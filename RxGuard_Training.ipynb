```python
import csv
import os

def reconstruct_labels(coco_json_path, images_folder):
    with open(coco_json_path) as f:
        data = json.load(f)

    cat_lookup = {c['id']: c['name'] for c in data['categories']}
    img_lookup = {img['id']: img['file_name'] for img in data['images']}

    # Group annotations by image_id
    from collections import defaultdict
    anns_by_image = defaultdict(list)
    for ann in data['annotations']:
        anns_by_image[ann['image_id']].append(ann)

    results = []
    for image_id, anns in anns_by_image.items():
        anns_sorted = sorted(anns, key=lambda a: a['bbox'][0])
        word = ''.join(cat_lookup[a['category_id']] for a in anns_sorted)
        filename = img_lookup[image_id]
        results.append((filename, word))

    return results

all_results = []
for split in ['train', 'valid', 'test']:
    split_folder = os.path.join(base, split)
    json_path = os.path.join(split_folder, "_annotations.coco.json")
    results = reconstruct_labels(json_path, split_folder)
    all_results.extend(results)
    print(f"{split}: {len(results)} labeled images")

# Save as CSV
csv_path = os.path.join(base, "reconstructed_labels.csv")
with open(csv_path, 'w', newline='') as f:
    writer = csv.writer(f)
    writer.writerow(['IMAGE', 'WORD'])
    writer.writerows(all_results)

print(f"\nSaved {len(all_results)} total labels to {csv_path}")
```

```python
import pandas as pd
import os

master_rows = []

# --- Kaggle Set 1: all three splits ---
for split, words_folder, labels_file in [
    ('Training', 'training_words', 'training_labels.csv'),
    ('Testing', 'testing_words', 'testing_labels.csv'),
    ('Validation', 'validation_words', 'validation_labels.csv'),
]:
    split_path = os.path.join(set1_base, split)
    labels_path = os.path.join(split_path, labels_file)
    images_path = os.path.join(split_path, words_folder)

    df = pd.read_csv(labels_path)
    print(f"{split} columns: {df.columns.tolist()}, rows: {len(df)}")

    for _, row in df.iterrows():
        image_full_path = os.path.join(images_path, row['IMAGE'])
        master_rows.append({
            'image_path': image_full_path,
            'text_label': row['MEDICINE_NAME'],
            'source': f'kaggle_set1_{split.lower()}'
        })

# --- Roboflow Set ---
roboflow_csv = os.path.join(dst3, "reconstructed_labels.csv")
rf_df = pd.read_csv(roboflow_csv)
print(f"Roboflow columns: {rf_df.columns.tolist()}, rows: {len(rf_df)}")

# Need to find which split subfolder each image actually lives in
for split in ['train', 'valid', 'test']:
    split_folder = os.path.join(dst3, split)
    if os.path.exists(split_folder):
        existing_files = set(os.listdir(split_folder))
        for _, row in rf_df.iterrows():
            if row['IMAGE'] in existing_files:
                master_rows.append({
                    'image_path': os.path.join(split_folder, row['IMAGE']),
                    'text_label': row['WORD'],
                    'source': f'roboflow_{split}'
                })

# Build master dataframe
master_df = pd.DataFrame(master_rows)
print(f"\nTotal merged rows: {len(master_df)}")
print(master_df['source'].value_counts())

# Save it
master_csv_path = os.path.join(drive_datasets, 'master_labels.csv')
master_df.to_csv(master_csv_path, index=False)
print(f"\nSaved master CSV to {master_csv_path}")
print(master_df.head())
```

```python
import cv2
import numpy as np
from PIL import Image
import os

processed_folder = os.path.join(drive_datasets, 'processed_images')
os.makedirs(processed_folder, exist_ok=True)

def preprocess_image(image_path, output_path):
    # Load with PIL first to handle mode conversion safely (RGBA, L, etc.)
    pil_img = Image.open(image_path).convert('RGB')
    img = np.array(pil_img)

    # Convert to grayscale for denoising/contrast, then back to RGB
    gray = cv2.cvtColor(img, cv2.COLOR_RGB2GRAY)

    # Denoise
    denoised = cv2.fastNlMeansDenoising(gray, h=10)

    # Contrast enhancement (CLAHE works well for uneven handwriting contrast)
    clahe = cv2.createCLAHE(clipLimit=2.0, tileGridSize=(8, 8))
    enhanced = clahe.apply(denoised)

    # Convert back to RGB (TrOCR expects 3-channel input)
    final = cv2.cvtColor(enhanced, cv2.COLOR_GRAY2RGB)

    Image.fromarray(final).save(output_path)

# Test on just 5 images first before running on all 5000+
sample_rows = master_df.sample(5, random_state=42)

for idx, row in sample_rows.iterrows():
    src = row['image_path']
    out_name = f"test_{idx}.png"
    out_path = os.path.join(processed_folder, out_name)
    preprocess_image(src, out_path)
    print(f"Processed: {out_name}")

print("\nTest batch done — check the processed_images folder in Drive to visually inspect results")
```

```python
from tqdm import tqdm

processed_paths = []
failed = []

for idx, row in tqdm(master_df.iterrows(), total=len(master_df)):
    src = row['image_path']
    out_name = f"img_{idx}.png"
    out_path = os.path.join(processed_folder, out_name)

    try:
        preprocess_image(src, out_path)
        processed_paths.append(out_path)
    except Exception as e:
        processed_paths.append(None)
        failed.append((src, str(e)))

master_df['processed_path'] = processed_paths

print(f"Successfully processed: {len(master_df) - len(failed)}")
print(f"Failed: {len(failed)}")
if failed:
    print("\nFirst few failures:")
    for path, err in failed[:5]:
        print(f"  {path}: {err}")

# Save updated master CSV with the new processed_path column
master_df.to_csv(master_csv_path, index=False)
print(f"\nUpdated master CSV saved with processed_path column")
```

```python
from transformers import ViTImageProcessor, RobertaTokenizer, TrOCRProcessor, VisionEncoderDecoderModel

image_processor = ViTImageProcessor.from_pretrained('microsoft/trocr-base-handwritten')
tokenizer = RobertaTokenizer.from_pretrained('microsoft/trocr-base-handwritten')
processor = TrOCRProcessor(image_processor=image_processor, tokenizer=tokenizer)

model = VisionEncoderDecoderModel.from_pretrained('microsoft/trocr-base-handwritten')

print("Model loaded successfully")
```

```python
from jiwer import cer, wer
from PIL import Image

model.eval()

sample = master_df.sample(30, random_state=42).reset_index(drop=True)

predictions = []
ground_truths = []

for idx, row in sample.iterrows():
    img = Image.open(row['processed_path']).convert('RGB')
    pixel_values = processor(images=img, return_tensors="pt").pixel_values.to('cuda')

    with torch.no_grad():
        generated_ids = model.generate(pixel_values)

    predicted_text = processor.batch_decode(generated_ids, skip_special_tokens=True)[0]

    predictions.append(predicted_text)
    ground_truths.append(row['text_label'])

    print(f"True: '{row['text_label']}'  |  Predicted: '{predicted_text}'")

overall_cer = cer(ground_truths, predictions)
overall_wer = wer(ground_truths, predictions)

print(f"\n--- Baseline Results (no fine-tuning) ---")
print(f"Character Error Rate: {overall_cer:.2%}")
print(f"Word Error Rate: {overall_wer:.2%}")
```
